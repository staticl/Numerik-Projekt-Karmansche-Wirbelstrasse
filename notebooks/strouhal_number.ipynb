```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
dt = 0.012 # 0.038 for n = m = 160, 0.012 for n = m = 256
```

```python
Re = 100 # Reynolds number, working point: 100 - 150

n = 256 # grid points in ξ-direction
m = 256 # grid points in θ-direction

simulation_time = 180 # total simulation time in seconds

simulation = VortexSimulation(Re, n, m, dt=dt)
omega, psi = simulation.run_simulation(simulation_time, steps_per_frame=5, to_plot=False)
```

```python
start_idx = int((simulation_time // dt) // 2)
start_idx
```

```python
signal = np.array(omega[start_idx:]) - np.mean(omega[start_idx:])  # Mittelwert abziehen
freqs = np.fft.rfftfreq(len(signal), d=dt)
spectrum = np.abs(np.fft.rfft(signal))

f_peak = freqs[np.argmax(spectrum[1:]) + 1]  # DC-Anteil (Index 0) ausschließen
St = 2 * f_peak
```

```python
theoretical_St = 0.198 * (1 - 19.7 / Re)
```

```python
diff = np.abs(St - theoretical_St) / theoretical_St
print(f"Relative Error: {diff:.4f}")
```
